```python
from collections import deque
import heapq




graph = {
    "Arad": {
        "Zerind": 75,
        "Sibiu": 140,
        "Timisoara": 118,
        "Oradea": 151
    },

    "Zerind": {
        "Arad": 75,
        "Oradea": 71
    },

    "Oradea": {
        "Zerind": 71,
        "Arad": 151
    },

    "Timisoara": {
        "Arad": 118,
        "Lugoj": 111
    },

    "Lugoj": {
        "Timisoara": 111,
        "Mehadia": 70
    },

    "Mehadia": {
        "Lugoj": 70,
        "Drobeta": 75
    },

    "Drobeta": {
        "Mehadia": 75,
        "Craiova": 120
    },

    "Craiova": {
        "Drobeta": 120,
        "Rimnicu Vilcea": 146,
        "Pitesti": 138
    },

    "Sibiu": {
        "Arad": 140,
        "Oradea": 151,
        "Rimnicu Vilcea": 80,
        "Fagaras": 99
    },

    "Rimnicu Vilcea": {
        "Sibiu": 80,
        "Craiova": 146,
        "Pitesti": 97
    },

    "Pitesti": {
        "Rimnicu Vilcea": 97,
        "Craiova": 138,
        "Bucharest": 101
    },

    "Fagaras": {
        "Sibiu": 99,
        "Bucharest": 211
    },

    "Bucharest": {
        "Fagaras": 211,
        "Pitesti": 101,
        "Giurgiu": 90,
        "Urziceni": 85
    },

    "Giurgiu": {
        "Bucharest": 90
    },

    "Urziceni": {
        "Bucharest": 85,
        "Vaslui": 142,
        "Hirsova": 98
    },

    "Vaslui": {
        "Urziceni": 142,
        "Iasi": 92
    },

    "Iasi": {
        "Vaslui": 92,
        "Neamt": 87
    },

    "Neamt": {
        "Iasi": 87
    },

    "Hirsova": {
        "Urziceni": 98,
        "Eforie": 86
    },

    "Eforie": {
        "Hirsova": 86
    }
}


START = "Arad"
GOAL = "Vaslui"




def show_result(name, path, cost):
    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    if path:
        print("Path :", " -> ".join(path))
        print("Cost :", cost)
        print("Nodes:", len(path))
    else:
        print("ไม่พบเส้นทาง")


def path_cost(path):
    cost = 0

    for i in range(len(path) - 1):
        cost += graph[path[i]][path[i + 1]]

    return cost



# 1. Breadth First Tree Search


def breadth_first_tree_search(start, goal):

    queue = [(start, [start])]

    while queue:

        node, path = queue.pop(0)

        if node == goal:
            return path, path_cost(path)

        for neighbor in graph[node]:

            # Tree Search สามารถสร้าง node ซ้ำได้
            new_path = path + [neighbor]

            queue.append((neighbor, new_path))

    return None, 0



# 2. Depth First Tree Search


def depth_first_tree_search(start, goal):

    stack = [(start, [start])]

    while stack:

        node, path = stack.pop()

        if node == goal:
            return path, path_cost(path)

        for neighbor in reversed(list(graph[node].keys())):

            # ป้องกันการวนกลับใน path เดียวกัน
            if neighbor not in path:
                stack.append(
                    (neighbor, path + [neighbor])
                )

    return None, 0



# 3. Breadth First Graph Search


def breadth_first_graph_search(start, goal):

    queue = deque([(start, [start])])
    visited = {start}

    while queue:

        node, path = queue.popleft()

        if node == goal:
            return path, path_cost(path)

        for neighbor in graph[node]:

            if neighbor not in visited:

                visited.add(neighbor)

                queue.append(
                    (neighbor, path + [neighbor])
                )

    return None, 0



# 4. Depth First Graph Search


def depth_first_graph_search(start, goal):

    stack = [(start, [start])]
    visited = set()

    while stack:

        node, path = stack.pop()

        if node in visited:
            continue

        visited.add(node)

        if node == goal:
            return path, path_cost(path)

        for neighbor in reversed(list(graph[node].keys())):

            if neighbor not in visited:
                stack.append(
                    (neighbor, path + [neighbor])
                )

    return None, 0



# 5. Best First Graph Search


def best_first_graph_search(start, goal, heuristic):

    queue = []

    heapq.heappush(
        queue,
        (heuristic[start], start, [start])
    )

    visited = set()

    while queue:

        h, node, path = heapq.heappop(queue)

        if node in visited:
            continue

        visited.add(node)

        if node == goal:
            return path, path_cost(path)

        for neighbor in graph[node]:

            if neighbor not in visited:

                heapq.heappush(
                    queue,
                    (
                        heuristic[neighbor],
                        neighbor,
                        path + [neighbor]
                    )
                )

    return None, 0



# 6. Uniform Cost Search


def uniform_cost_search(start, goal):

    queue = []

    heapq.heappush(
        queue,
        (0, start, [start])
    )

    visited = {}

    while queue:

        cost, node, path = heapq.heappop(queue)

        if node in visited:
            continue

        visited[node] = cost

        if node == goal:
            return path, cost

        for neighbor, distance in graph[node].items():

            if neighbor not in visited:

                heapq.heappush(
                    queue,
                    (
                        cost + distance,
                        neighbor,
                        path + [neighbor]
                    )
                )

    return None, 0


# 7. Depth Limited Search


def depth_limited_search(start, goal, limit):

    def recursive_dls(node, path, depth):

        if node == goal:
            return path

        if depth == 0:
            return None

        for neighbor in graph[node]:

            if neighbor not in path:

                result = recursive_dls(
                    neighbor,
                    path + [neighbor],
                    depth - 1
                )

                if result:
                    return result

        return None

    path = recursive_dls(start, [start], limit)

    if path:
        return path, path_cost(path)

    return None, 0

# 8. Iterative Deepening Search


def iterative_deepening_search(start, goal, max_depth=30):

    for depth in range(max_depth + 1):

        path, cost = depth_limited_search(
            start,
            goal,
            depth
        )

        if path:
            return path, cost

    return None, 0



# Heuristic
# ค่า heuristic โดยประมาณไปยัง Vaslui
# ใช้สำหรับ Best First, A* และ Recursive Best First

heuristic = {
    "Arad": 350,
    "Zerind": 330,
    "Oradea": 390,
    "Timisoara": 370,
    "Lugoj": 300,
    "Mehadia": 250,
    "Drobeta": 210,
    "Craiova": 200,
    "Sibiu": 280,
    "Rimnicu Vilcea": 220,
    "Fagaras": 250,
    "Pitesti": 160,
    "Bucharest": 227,
    "Giurgiu": 290,
    "Urziceni": 142,
    "Hirsova": 200,
    "Eforie": 280,
    "Vaslui": 0,
    "Iasi": 92,
    "Neamt": 170
}


# 10. A* Search


def a_star_search(start, goal, heuristic):

    queue = []

    heapq.heappush(
        queue,
        (
            heuristic[start],
            0,
            start,
            [start]
        )
    )

    best_cost = {}

    while queue:

        f, g, node, path = heapq.heappop(queue)

        if node in best_cost and g > best_cost[node]:
            continue

        best_cost[node] = g

        if node == goal:
            return path, g

        for neighbor, distance in graph[node].items():

            new_g = g + distance
            new_f = new_g + heuristic[neighbor]

            if (
                neighbor not in best_cost
                or new_g < best_cost[neighbor]
            ):

                heapq.heappush(
                    queue,
                    (
                        new_f,
                        new_g,
                        neighbor,
                        path + [neighbor]
                    )
                )

    return None, 0


# 11. Recursive Best First Search (RBFS)


def recursive_best_first_search(start, goal, heuristic):

    def rbfs(node, path, g, f_limit):

        if node == goal:
            return path, g, f_limit

        successors = []

        for neighbor, distance in graph[node].items():

            if neighbor not in path:

                new_g = g + distance
                f = max(
                    new_g + heuristic[neighbor],
                    g + heuristic[node]
                )

                successors.append(
                    [f, neighbor, new_g]
                )

        if not successors:
            return None, float("inf"), f_limit

        while True:

            successors.sort(
                key=lambda x: x[0]
            )

            best = successors[0]

            if best[0] > f_limit:
                return None, best[0], f_limit

            alternative = (
                successors[1][0]
                if len(successors) > 1
                else float("inf")
            )

            result, new_f, result_limit = rbfs(
                best[1],
                path + [best[1]],
                best[2],
                min(f_limit, alternative)
            )

            if result:
                return result, new_f, result_limit

            best[0] = new_f

    path, cost, _ = rbfs(
        start,
        [start],
        0,
        float("inf")
    )

    if path:
        return path, cost

    return None, 0

# 9. Greedy Best First Search


def greedy_best_first_search(start, goal, heuristic):

    queue = []

    heapq.heappush(
        queue,
        (
            heuristic[start],
            start,
            [start]
        )
    )

    visited = set()

    while queue:

        h, node, path = heapq.heappop(queue)

        if node in visited:
            continue

        visited.add(node)

        if node == goal:
            return path, path_cost(path)

        for neighbor in graph[node]:

            if neighbor not in visited:

                heapq.heappush(
                    queue,
                    (
                        heuristic[neighbor],
                        neighbor,
                        path + [neighbor]
                    )
                )

    return None, 0



path, cost = breadth_first_tree_search(START, GOAL)
show_result("1. Breadth First Tree Search", path, cost)

path, cost = depth_first_tree_search(START, GOAL)
show_result("2. Depth First Tree Search", path, cost)

path, cost = breadth_first_graph_search(START, GOAL)
show_result("3. Breadth First Graph Search", path, cost)

path, cost = depth_first_graph_search(START, GOAL)
show_result("4. Depth First Graph Search", path, cost)

path, cost = best_first_graph_search(
    START,
    GOAL,
    heuristic
)
show_result("5. Best First Graph Search", path, cost)

path, cost = uniform_cost_search(START, GOAL)
show_result("6. Uniform Cost Search", path, cost)

path, cost = depth_limited_search(
    START,
    GOAL,
    limit=10
)
show_result("7. Depth Limited Search", path, cost)

path, cost = iterative_deepening_search(
    START,
    GOAL
)
show_result("8. Iterative Deepening Search", path, cost)

path, cost = greedy_best_first_search(
    START,
    GOAL,
    heuristic
)
show_result("9. Greedy Best First Search", path, cost)

path, cost = a_star_search(
    START,
    GOAL,
    heuristic
)
show_result("10. A* Search", path, cost)

path, cost = recursive_best_first_search(
    START,
    GOAL,
    heuristic
)
show_result("11. Recursive Best First Search", path, cost)
```
